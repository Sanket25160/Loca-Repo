#Importing Libraries

```python
from typing import TypedDict
from langgraph.graph import StateGraph, END
import re
```

#Define Agent State

```python
class ExpenseState(TypedDict):
  user_input: str
  amount: str
  decision: str
  result: str
```

#Planner Node

```python
def planner(state: ExpenseState):
  text = state["user_input"]

  match = re.search(r"\d+", text)
  amount = int(match.group()) if match else 0

  return {
      "amount" : amount,
      "decision" : "check_policy",
      "retries" : state.get("retries", 0)
  }
```

#Policy Node

```python
def policy_tool(state: ExpenseState):
  EXPENSE_LIMIT = 2000

  if state["amount"] <= EXPENSE_LIMIT:
    return {"decision" : "approved"}
  else:
    return {"decision" : "rejected"}
```

#Validator Node

```python
def validator(state: ExpenseState):
  if state["decision"]=="approved":
    return {"decision" : "approved"}
  if state["retries"] >=2:
    return{"decision": "stop"}

  return{
      "decision": "retry",
      "retries" : state["retries"]+1
  }
```

#Final Answer Node

```python
def final_answer(state: ExpenseState):
  if state["decision"]=="approved":
    return {
        "result" : f"Expense of {state["amount"]} is eligible as per policy we will reimbuse this"
    }
  else:
    return{
        "result" : f"Expense of {state["amount"]} is not eligible as per policy so we will not reimbuse this"
    }
```

#Create Langgraph

```python
graph = StateGraph(ExpenseState)

graph.add_node("planner", planner)
graph.add_node("Policy Check", policy_tool)
graph.add_node("Validator", validator)
graph.add_node("Final", final_answer)

graph.set_entry_point("planner")
graph.add_edge("planner", "Policy Check")
graph.add_edge("Policy Check", "Validator")

graph.add_conditional_edges(
    "Validator",
    lambda state: state["decision"],
    {
        "approved" : "Final",
        "retry" : "planner",
        "stop" : "Final"
    }
)
graph.add_edge("Final", END)

expense_agent = graph.compile()
```

#Run this graph

```python
user_text = input("Enter your expense")

output = expense_agent.invoke({
    "user_input": user_text
})

print("\n Agent Response")
print(output["result"])
```
